```python
import pandas as pd

# Tải dữ liệu returns.csv
returns_df = pd.read_csv('/content/returns.csv')

# Hiển thị thông tin tổng quan và vài dòng dữ liệu mẫu
print("--- THÔNG TIN TỔNG QUAN ---")
returns_df.info()

print("\n--- 5 DÒNG ĐẦU TIÊN ---")
display(returns_df.head())

# Kiểm tra giá trị bị khuyết (NaN/Null)
print("\n--- SỐ LƯỢNG GIÁ TRỊ NULL TRÊN MỖI CỘT ---")
print(returns_df.isnull().sum())

# Kiểm tra số lượng dòng trùng lặp hoàn toàn
print("\n--- SỐ DÒNG TRÙNG LẶP HOÀN TOÀN ---")
print(returns_df.duplicated().sum())
```

```python
# 1. Tạo bảng danh mục Return Reasons
unique_reasons = returns_df['return_reason'].unique()
reasons_df = pd.DataFrame({
    'reason_id': [f'REA-{i+1:02d}' for i in range(len(unique_reasons))],
    'return_reason': unique_reasons
})

# 2. Thay thế 'return_reason' trong bảng gốc bằng 'reason_id'
returns_3nf = returns_df.merge(reasons_df, on='return_reason', how='left')
returns_3nf = returns_3nf.drop(columns=['return_reason'])

# Sắp xếp lại thứ tự cột cho đẹp mắt
returns_3nf = returns_3nf[['return_id', 'order_id', 'product_id', 'return_date', 'reason_id', 'return_quantity', 'refund_amount']]

# 3. Xuất hai bảng đã chuẩn hóa ra file .csv mới
returns_3nf.to_csv('/content/returns_3nf.csv', index=False)
reasons_df.to_csv('/content/return_reasons_dim.csv', index=False)

# Hiển thị kết quả kiểm tra
print("--- BẢNG DANH MỤC LÝ DO (return_reasons_dim) ---")
display(reasons_df)

print("\n--- 5 DÒNG ĐẦU BẢNG GIAO DỊCH TRẢ HÀNG 3NF (returns_3nf) ---")
display(returns_3nf.head())

print("\nĐã xuất thành công 2 file mới:\n1. /content/returns_3nf.csv\n2. /content/return_reasons_dim.csv")
```
