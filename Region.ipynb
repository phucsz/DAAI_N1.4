```python
import pandas as pd
import numpy as np

# Đọc tệp geography.csv
geo_df = pd.read_csv('/content/geography.csv')

# 1. Trích xuất cột 'region' và loại bỏ các giá trị null/NaN
region_series = geo_df['region'].dropna()

# 2. Loại bỏ các giá trị trùng lặp và sắp xếp lại cho đẹp mắt
unique_regions = region_series.drop_duplicates().sort_values().reset_index(drop=True)

# 3. Tạo DataFrame mới chứa region và region_id
region_df = pd.DataFrame({
    'region_id': [f'REG_{i+1:02d}' for i in range(len(unique_regions))],
    'region': unique_regions
})

# 4. Lưu kết quả ra file mới tên là region.csv
output_path = '/content/region.csv'
region_df.to_csv(output_path, index=False)

# Hiển thị thông tin kết quả thu được
print(f"Đã tạo thành công bảng region tại: {output_path}")
print(f"Tổng số dòng: {len(region_df)}")
display(region_df)
```
