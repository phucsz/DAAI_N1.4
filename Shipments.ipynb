### 3. Chuẩn hóa bảng Giao dịch Vận chuyển (Shipments/Delivery Transactions)
Trích xuất các cột liên quan trực tiếp đến giao dịch vận chuyển: `order_id`, `shipper_id`, `ship_date`, `delivery_date`, `shipping_fee` để tạo bảng liên kết chuẩn 3NF.

```python
# Lọc các cột phục vụ bảng giao dịch vận chuyển
shipment_trans_cols = ['order_id', 'shipper_id', 'ship_date', 'delivery_date', 'shipping_fee']

# Kiểm tra sự tồn tại của cột trong dataframe gốc để tránh lỗi
available_trans_cols = [col for col in shipment_trans_cols if col in shipments_df.columns]
if 'shipping_fee' not in available_trans_cols and 'shipping_feee' in shipments_df.columns:
    # Đề phòng trường hợp cột bị sai chính tả trong file gốc
    shipments_df = shipments_df.rename(columns={'shipping_feee': 'shipping_fee'})
    available_trans_cols.append('shipping_fee')

# Tạo dataframe
shipment_trans_df = shipments_df[available_trans_cols].copy()

# 1. Kiểm tra và xử lý dữ liệu khuyết thiếu (Null/NaN)
print("--- SỐ LƯỢNG GIÁ TRỊ NULL TRONG CÁC CỘT GIAO DỊCH ---")
print(shipment_trans_df.isnull().sum())

# Điền khuyết thiếu nếu có
for col in shipment_trans_df.columns:
    if shipment_trans_df[col].isnull().any():
        if shipment_trans_df[col].dtype in ['int64', 'float64']:
            shipment_trans_df[col] = shipment_trans_df[col].fillna(0)
        else:
            shipment_trans_df[col] = shipment_trans_df[col].fillna('Unknown')

# 2. Loại bỏ dữ liệu trùng lặp hoàn toàn
initial_rows = len(shipment_trans_df)
shipment_trans_df = shipment_trans_df.drop_duplicates().reset_index(drop=True)
print(f"\nĐã loại bỏ {initial_rows - len(shipment_trans_df)} dòng trùng lặp hoàn toàn.")

# 3. Xuất dữ liệu ra file CSV mới
shipment_trans_df.to_csv('/content/shipments_3nf.csv', index=False)
print("\nĐã lưu file thành công tại: /content/shipments_3nf.csv")

# Hiển thị cấu trúc và dữ liệu mẫu
print("\n--- THÔNG TIN CHI TIẾT BẢNG SHIPMENTS 3NF ---")
shipment_trans_df.info()
display(shipment_trans_df.head())
```
