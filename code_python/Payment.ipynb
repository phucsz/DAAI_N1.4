```python
import pandas as pd

# Tải tệp payments.csv
payments_df = pd.read_csv('/content/payments.csv')

# Hiển thị 5 dòng đầu tiên của DataFrame
display(payments_df.head())
```

```python
# Hiển thị thông tin tổng quan về DataFrame, bao gồm kiểu dữ liệu và giá trị thiếu
payments_df.info()
```

```python
# Kiểm tra các giá trị duy nhất trong cột 'payment_method'
print("Các phương thức thanh toán duy nhất:")
print(payments_df['payment_method'].unique())

# Kiểm tra thống kê mô tả cho 'payment_value' và 'installments'
print("\nThống kê mô tả cho 'payment_value':")
print(payments_df['payment_value'].describe())

print("\nThống kê mô tả cho 'installments':")
print(payments_df['installments'].describe())
```

```python
# Kiểm tra các dòng trùng lặp hoàn toàn
print("\nSố lượng dòng trùng lặp hoàn toàn:", payments_df.duplicated().sum())

# Kiểm tra xem order_id có phải là duy nhất không (mỗi order có thể có nhiều payment, nhưng mỗi dòng nên đại diện cho một payment duy nhất)
# Nếu mỗi order_id chỉ có một phương thức thanh toán, thì order_id có thể là PK.
# Nếu một order_id có nhiều phương thức thanh toán, thì cần một PK tổng hợp hoặc Payment_ID riêng.
print("\nSố lượng order_id duy nhất:", payments_df['order_id'].nunique())
print("Tổng số lượng dòng:", len(payments_df))
```

```python
# Lưu DataFrame đã được kiểm tra (và xem như đã chuẩn hóa) vào một tệp CSV mới
output_filename = 'payments_3nf_cleaned.csv'
payments_df.to_csv(output_filename, index=False)

print(f"Dữ liệu đã được lưu vào '{output_filename}' thành công.")
```
