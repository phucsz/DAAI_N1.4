```python
import pandas as pd

# Tải dữ liệu từ reviews.csv
reviews_df = pd.read_csv('/content/reviews.csv')

# Hiển thị thông tin cấu trúc tổng quan
print("--- THÔNG TIN CẤU TRÚC BẢNG REVIEWS ---")
reviews_df.info()

# Hiển thị vài mẫu dữ liệu đầu tiên
print("\n--- 5 DÒNG DỮ LIỆU ĐẦU TIÊN ---")
display(reviews_df.head())

# Kiểm tra giá trị Null/NaN
print("\n--- SỐ LƯỢNG GIÁ TRỊ NULL TRONG MỖI CỘT ---")
print(reviews_df.isnull().sum())

# Kiểm tra số dòng trùng lặp
print("\n--- SỐ DÒNG BỊ TRÙNG LẶP HOÀN TOÀN ---")
print(reviews_df.duplicated().sum())
```

```python
# 1. Tạo bảng danh mục Review Titles để chuẩn hóa 3NF
unique_titles = reviews_df['review_title'].unique()
review_titles_df = pd.DataFrame({
    'title_id': [f'TIL-{i+1:03d}' for i in range(len(unique_titles))],
    'review_title': unique_titles
})

# 2. Ánh xạ khóa ngoại 'title_id' vào bảng chính và loại bỏ cột chuỗi cũ
reviews_3nf = reviews_df.merge(review_titles_df, on='review_title', how='left')
reviews_3nf = reviews_3nf.drop(columns=['review_title'])

# Sắp xếp lại thứ tự cột hợp lý
reviews_3nf = reviews_3nf[['review_id', 'order_id', 'product_id', 'customer_id', 'review_date', 'rating', 'title_id']]

# 3. Xuất kết quả ra các file .csv mới
reviews_3nf.to_csv('/content/reviews_3nf.csv', index=False)
review_titles_df.to_csv('/content/review_titles_dim.csv', index=False)

# Hiển thị vài dòng mẫu kiểm tra
print("--- BẢNG DANH MỤC TIÊU ĐỀ ĐÁNH GIÁ (review_titles_dim) ---")
display(review_titles_df.head())

print("\n--- 5 DÒNG ĐẦU BẢNG ĐÁNH GIÁ ĐÃ CHUẨN HÓA 3NF (reviews_3nf) ---")
display(reviews_3nf.head())

print("\nĐã xuất thành công 2 file mới:\n1. /content/reviews_3nf.csv\n2. /content/review_titles_dim.csv")
```
